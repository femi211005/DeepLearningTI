```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
from tensorflow.keras.preprocessing import sequence
import matplotlib.pyplot as plt

print(f"TensorFlow Version: {tf.__version__}")
```

##CNN

```python
# 1. Load & Preprocess Dataset IMDb
max_features = 10000
maxlen = 200

print("Memuat dataset IMDb...")
(X_train, y_train), (X_test, y_test) = datasets.imdb.load_data(num_words=max_features)

# Menyamakan panjang urutan teks (padding)
X_train = sequence.pad_sequences(X_train, maxlen=maxlen)
X_test = sequence.pad_sequences(X_test, maxlen=maxlen)
```

```python
# 2. Membangun Arsitektur 1D CNN untuk Teks
cnn_model = models.Sequential([
    layers.Embedding(max_features, 32, input_length=maxlen),
    layers.Conv1D(64, 5, activation='relu'),
    layers.MaxPooling1D(pool_size=4),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(1, activation='sigmoid')
])

cnn_model.summary()
```

```python
# 3. Kompilasi & Training Model CNN
cnn_model.compile(optimizer='adam',
                  loss='binary_crossentropy',
                  metrics=['accuracy'])

print("\nMemulai Training Model CNN...")
history_cnn = cnn_model.fit(X_train, y_train,
                            epochs=5,
                            batch_size=64,
                            validation_data=(X_test, y_test))
```

```python
# 4. Evaluasi Model CNN
test_loss_cnn, test_acc_cnn = cnn_model.evaluate(X_test, y_test, verbose=2)
print(f"\nAkurasi Model CNN pada Data Testing: {test_acc_cnn * 100:.2f}%")

# Simpan model CNN
cnn_model.save('imdb_cnn_model.keras')
print("Model CNN berhasil disimpan!")
```

##RNN (LSTM)

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
```

```python
# 1. Membangun Arsitektur RNN (LSTM)
# (Dataset X_train, X_test, y_train, y_test sudah otomatis tersedia dari cell CNN sebelumnya)
rnn_model = models.Sequential([
    layers.Embedding(max_features, 32, input_length=maxlen),
    layers.LSTM(64, dropout=0.2, recurrent_dropout=0.2),
    layers.Dense(1, activation='sigmoid')
])

rnn_model.summary()
```

```python
# 2. Kompilasi & Training Model RNN
rnn_model.compile(optimizer='adam',
                  loss='binary_crossentropy',
                  metrics=['accuracy'])

print("\nMemulai Training Model RNN (LSTM)...")
history_rnn = rnn_model.fit(X_train, y_train,
                            epochs=5,
                            batch_size=64,
                            validation_data=(X_test, y_test))
```

```python
# 3. Evaluasi Model RNN
test_loss_rnn, test_acc_rnn = rnn_model.evaluate(X_test, y_test, verbose=2)
print(f"\nAkurasi Model RNN pada Data Testing: {test_acc_rnn * 100:.2f}%")

# Simpan model RNN
rnn_model.save('imdb_rnn_model.keras')
print("Model RNN berhasil disimpan!")
```

```python
# 4. KESIMPULAN PERBANDINGAN AKHIR
print("HASIL PERBANDINGAN ARSITEKTUR PADA 1 DATASET (IMDb):")
print(f"1. Akurasi 1D CNN : {test_acc_cnn * 100:.2f}%")
print(f"2. Akurasi RNN-LSTM: {test_acc_rnn * 100:.2f}%")
```

## 📌 Kesimpulan Perbandingan Model Deep Learning

Berdasarkan implementasi, pelatihan, dan pengujian yang telah dilakukan menggunakan satu dataset teks yang sama (*IMDb Reviews*), berikut adalah kesimpulan dari kedua arsitektur Deep Learning yang diuji:

**1. 1D CNN (Convolutional Neural Network):**
- **Fokus & Mekanisme:** Menggunakan lapisan konvolusi 1 dimensi (*Conv1D*) dan *pooling* untuk mengekstraksi pola lokal atau frasa kata penting secara hierarkis dari teks ulasan.
- **Karakteristik:** Proses pelatihan (*training*) cenderung lebih cepat per *epoch*-nya karena sifat komputasi paralel dari operasi konvolusi.
- **Hasil:** Mampu mengenali pola kata kunci sentimen dengan baik dan menghasilkan tingkat akurasi yang kompetitif pada data pengujian (*testing*).

**2. RNN - LSTM (Recurrent Neural Network):**
- **Fokus & Mekanisme:** Dirancang secara khusus untuk data sekuensial dengan memanfaatkan sel memori (*hidden state/LSTM*) yang mampu mempertahankan konteks, urutan kata, dan hubungan jangka panjang dalam sebuah kalimat.
- **Karakteristik:** Membutuhkan waktu komputasi yang sedikit lebih lama per *epoch* karena pemrosesan data dilakukan secara berurutan (*sequential*).
- **Hasil:** Sangat andal dalam memahami struktur kalimat yang kompleks serta konteks keseluruhan teks untuk melakukan analisis sentimen secara akurat.

**Kesimpulan Akhir Tugas:**
Melalui perbandingan langsung pada dataset teks *IMDb Reviews*, dapat disimpulkan bahwa kedua arsitektur (CNN dan RNN) sama-sama berhasil diterapkan untuk tugas analisis sentimen. **RNN (LSTM)** umumnya unggul dalam memahami urutan konteks kalimat secara utuh karena sifat sekuensialnya, sementara **1D CNN** menawarkan keunggulan dalam kecepatan komputasi dan efisiensi penangkapan fitur lokal. Pemilihan arsitektur dapat disesuaikan dengan kebutuhan prioritas antara kecepatan pelatihan atau pemahaman konteks urutan teks yang mendalam.
